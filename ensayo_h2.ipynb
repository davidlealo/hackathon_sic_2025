```python
import requests
from bs4 import BeautifulSoup

# URL de la página de Samsung en Wikipedia
url = "https://es.wikipedia.org/wiki/Samsung"

try:
    # Definir un User-Agent para simular un navegador
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
    }

    # Realizar la solicitud HTTP GET con los headers
    response = requests.get(url, headers=headers)
    response.raise_for_status()  # Lanzar una excepción para códigos de estado de error (4xx o 5xx)

    # Parsear el contenido HTML con BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    print(f"--- Scraping de: {url} ---")

    # Extraer el título de la página
    title = soup.find('h1', id='firstHeading')
    if title:
        print(f"Título de la página: {title.text.strip()}")
    else:
        print("No se pudo encontrar el título de la página.")

    # Extraer el primer párrafo del contenido principal
    first_paragraph = soup.find('div', class_='mw-parser-output').find('p')
    if first_paragraph:
        print("\nPrimer párrafo:")
        print(first_paragraph.text.strip())
    else:
        print("No se pudo encontrar el primer párrafo.")

except requests.exceptions.RequestException as e:
    print(f"Error al realizar la solicitud: {e}")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```
